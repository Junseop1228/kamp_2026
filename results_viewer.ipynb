# 결과 열람 노트북

`python run.py`가 만든 `results/`를 읽어 보여주기만 한다. 계산은 하지 않는다.

```python
import json
from pathlib import Path

import pandas as pd
from IPython.display import Image, display

R = Path('results')
LINES = ['cn7', 'rg3']
load = lambda line, name: json.loads((R / line / name).read_text(encoding='utf-8'))
```

## 1. 데이터 진단

```python
for line in LINES:
    d = load(line, 'diagnose_summary.json')
    print(line, d['imbalance']['train'], '| 쌍 비율 train/test',
          round(d['groups']['train']['rows_in_multi_share'], 3), round(d['groups']['test']['rows_in_multi_share'], 3),
          '| 하한/상한 F1', round(d['f1_floor'], 4), round(d['f1_ceiling'], 4))
display(Image(R / 'figures' / 'fig1_pair_share.png'))
```

## 2. 모델 비교와 선택

```python
for line in LINES:
    print(line, load(line, 'model_summary.json')['selected_model'])
    display(pd.read_csv(R / line / 'selection.csv').round(4))
    display(Image(R / 'figures' / f'fig2_model_f1_{line}.png'))
```

## 3. 영향요인과 오류분석

```python
for line in LINES:
    print(line, load(line, 'analysis_summary.json')['fn_type_counts'])
    display(pd.read_csv(R / line / 'importance.csv').head(8).round(3))
    display(pd.read_csv(R / line / 'interaction_2x2.csv'))
display(Image(R / 'figures' / 'fig3_fn_types.png'))
```

## 4. 검사 우선순위와 알람

```python
for line in LINES:
    display(pd.read_csv(R / line / 'budget_table.csv').round(3))
    print(line, load(line, 'prioritize_summary.json'))
display(Image(R / 'figures' / 'fig4_budget.png'))
```
